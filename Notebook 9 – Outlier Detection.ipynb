# Outlier Detection

An outlier is a data value that is very different from most of the other values in a dataset.

In simple words, an outlier is an unusually high or unusually low value compared with the rest of the data.
### Example
consider employee salaries:

25000, 28000, 30000, 32000, 35000, 38000, 40000, 250000

Most salaries are between ₹25,000 and ₹40,000, but ₹2,50,000 is much higher than the others. Therefore, ₹2,50,000 can be considered an outlier.

Outliers can occur because of:

- Data entry mistakes
- Measurement errors
- Unusual events
- Genuine extreme observations
- Fraud or abnormal behavior

```python
import numpy as np

salary = np.array([25000, 28000, 30000, 32000, 35000, 38000, 40000, 250000])

print("Salary values:", salary)
print("Mean:", np.mean(salary))
```

## Types of Outliers
### Global Outlier

A value that is very different from the entire dataset.

#### Example:

10, 12, 11, 13, 12, 100

### Contextual Outlier

A value that is unusual in a particular situation or context.

#### Example:

Temperature:

Chennai - Summer: 35°C

35°C is normal in summer, but 35°C might be unusual in some other context.
### Collective Outlier

A group of values that together behave unusually.

#### Example
A sudden sequence of unusual network traffic could indicate a cyberattack.

## Detecting Outliers

There are several methods for detecting outliers:

- IQR Method
- Z-Score Method
- Box Plot
- Scatter Plot

## IQR Method
QR (Interquartile Range) is a method used to detect outliers in a dataset.

It focuses on the middle 50% of the data and ignores extreme values.

The IQR is calculated as:

$$ IQR = Q3 - Q1 $$

Where:

- Q1 (First Quartile) → 25th percentile
- Q2 (Median) → 50th percentile
- Q3 (Third Quartile) → 75th percentile
  ### Example

```python
import numpy as np

# Employee salaries
salary = np.array([20, 22, 24, 25, 26, 28, 30, 32, 35, 100])

# Calculate Q1 and Q3
Q1 = np.percentile(salary, 25)
Q3 = np.percentile(salary, 75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = salary[(salary < lower_bound) | (salary > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Z-Score Method
Z-Score is a method used to identify outliers by checking how far a data value is from the mean.

It tells us the number of standard deviations a value is away from the mean.

The formula is:

$$ Z = \frac{X-\mu}{\sigma} $$

Where:

- X → Data value
- μ → Mean
- σ → Standard deviation
- Z → Z-score
  ## Example:

- Z = 0       → Value is exactly at the mean
- Z = +1      → 1 standard deviation above mean
- Z = -2      → 2 standard deviations below mean
- Z = +4      → Possible outlier
 
  [30000, 32000, 34000, 35000, 36000, 38000, 40000, 100000]

```python
import numpy as np

# Employee salaries
salary = np.array([30000, 32000, 34000, 35000, 36000, 38000, 40000, 100000])

# Calculate mean and standard deviation
mean = np.mean(salary)
std = np.std(salary)

# Calculate Z-scores
z_scores = (salary - mean) / std

print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-Scores:", z_scores)

# Identify outliers
outliers = salary[np.abs(z_scores) > 3]

print("Outliers:", outliers)
```

## Box PLot
A Box Plot (also called a Box-and-Whisker Plot) is a graph used to visualize the distribution of data and identify outliers.

A box plot shows five important values:

Minimum – Smallest value within the normal range
- Q1 – 25th percentile
- Median (Q2) – Middle value
- Q3 – 75th percentile
- Maximum – Largest value within the normal range

The IQR (Interquartile Range) is:

IQR=Q3−Q1


```python
import numpy as np
import matplotlib.pyplot as plt

# Employee salary data
salary = np.array([20, 22, 24, 25, 26, 28, 30, 32, 35, 100])

# Create box plot
plt.boxplot(salary)

plt.title("Employee Salary - Box Plot")
plt.ylabel("Salary")
plt.show()
```

## Scatter Plot
A Scatter Plot is a graph used to show the relationship between two numerical variables.

It represents each observation as a point (dot) on a graph.

- X-axis → First variable
- Y-axis → Second variable
- Each dot → One observation
### Example
We can use a scatter plot to understand the relationship between years of experience and salary.
#### Types of Relationships
1. Positive Relationship

When one variable increases, the other variable also tends to increase.

2. Negative Relationship

When one variable increases, the other tends to decrease.

3. No Relationship

There is no clear pattern between the variables


```python
import matplotlib.pyplot as plt

# Employee data
experience = [1, 3, 5, 8, 10, 15]
salary = [25000, 32000, 40000, 55000, 65000, 80000]

# Create scatter plot
plt.scatter(experience, salary)

plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```
