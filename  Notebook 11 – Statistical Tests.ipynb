# Statistical Tests
Statistical tests help us decide whether an observed difference or relationship in sample data is likely to be real or could have happened by chance.

## T-Test
Definition

A T-Test is a statistical test used to compare the means (averages) of groups and find out whether the difference between them is statistically significant.
### Example

Suppose a company wants to know whether a new training program improves employee productivity.

Average productivity before training = 70
Average productivity after training = 78

The company can use a T-Test to check whether the increase from 70 to 78 is statistically significant.

If the p-value < 0.05, we usually conclude that the training had a statistically significant effect.
##### Types of T-Test

There are mainly three types:

1. One-Sample T-Test

Compares a sample mean with a known or expected value.

##### Example:

A company expects employees to complete an average of 50 tasks per day.

We collect data from 20 employees and test whether their average is significantly different from 50.

2. Independent Two-Sample T-Test

Compares the means of two independent groups.

##### Example:

Compare the average sales of:

Team A
Team B

The employees in Team A and Team B are different people.

3. Paired T-Test

Compares two measurements from the same subjects.

##### Example:

Compare employee productivity:

Before training
After training

The same employees are measured twice.



```python
from scipy.stats import ttest_ind

# Accuracy of Model A
model_a = [85, 87, 86, 88, 84, 89, 87, 86]

# Accuracy of Model B
model_b = [80, 82, 81, 83, 79, 84, 82, 81]

# Independent T-Test
t_stat, p_value = ttest_ind(model_a, model_b)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

## Z-Test
A Z-Test is a statistical test used to determine whether there is a significant difference between a sample mean and a population mean, or between the means of two groups.
#### Formula

For a one-sample Z-Test:

$$ Z = \frac{\bar{x}-\mu}{\sigma/\sqrt{n}} $$

Where:

- \(\bar{x}\) = Sample mean
- \(\mu\) = Population mean
- \(\sigma\) = Population standard deviation
- \(n\) = Sample size
  #### Types of Z-Test
1. One-Sample Z-Test

Compares a sample mean with a known population mean.

##### Example:

Is the average weight of products different from 500 grams?

2. Two-Sample Z-Test

Compares the means of two independent groups.

##### Example:

Is the average salary of employees in Company A different from Company B?

```python
import numpy as np
from scipy.stats import norm

sample_mean = 32
population_mean = 30
population_std = 5
sample_size = 50

z_stat = (sample_mean - population_mean) / (
    population_std / np.sqrt(sample_size)
)

p_value = 2 * (1 - norm.cdf(abs(z_stat)))

print("Z-Statistic:", z_stat)
print("P-Value:", p_value)
```

## Chi-Square Test
The Chi-Square Test (χ² Test) is a statistical test used to determine whether there is a significant relationship or difference between categorical variables.
### Chi-Square Formula
$$ \chi^2 = \sum \frac{(O-E)^2}{E} $$

Where:

- O = Observed frequency
- E = Expected frequency
- Σ = Sum of all values
#### Types of Chi-Square Tests
1. Chi-Square Test of Independence

Used to check whether two categorical variables are related.

Example:

Is gender related to product preference?

2. Chi-Square Goodness-of-Fit Test

Used to check whether observed categorical data matches an expected distribution.

Example:

A company expects customers to choose four products equally:
Product A → 25%
Product B → 25%
Product C → 25%
Product D → 25%


```python
import numpy as np
from scipy.stats import chi2_contingency

data = np.array([
    [40, 30],
    [20, 50]
])

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square Statistic:", chi2)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)
```

## ANOVA
ANOVA (Analysis of Variance) is a statistical test used to compare the means of three or more groups and determine whether there is a statistically significant difference between them.
### Example

Suppose a company wants to compare the average productivity of employees using three different training methods:

Training A
Training B
Training C
### Types of ANOVA
1. One-Way ANOVA

Used when there is one categorical independent variable with three or more groups.

Example:

Compare employee productivity across three training methods.

2. Two-Way ANOVA

Used when there are two independent categorical variables.

Example:

Study the effect of training method and work location on employee productivity.

It can examine:

Effect of training
Effect of work location
Interaction between training and work location


```python
from scipy.stats import f_oneway

strategy_a = [20, 22, 19, 21, 23]
strategy_b = [30, 28, 32, 31, 29]
strategy_c = [25, 24, 26, 23, 27]

f_stat, p_value = f_oneway(
    strategy_a,
    strategy_b,
    strategy_c
)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)
```

## Mann-Whitney U Test
The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.
It is an alternative to the independent T-Test when the assumptions of the T-Test are not satisfied.
Instead of directly comparing means, it works with the ranks of the observations.
### Example
An e-commerce company wants to compare customer satisfaction scores between:
- Website A
- Website B

```python
from scipy.stats import mannwhitneyu

website_a = [2, 3, 4, 2, 5, 3]
website_b = [4, 5, 5, 4, 3, 5]

u_stat, p_value = mannwhitneyu(
    website_a,
    website_b,
    alternative="two-sided"
)

print("U-Statistic:", u_stat)
print("P-Value:", p_value)
```

## Shapiro-Wilk Test
The Shapiro-Wilk Test is a statistical test used to check whether a dataset is normally distributed.
It is commonly used before statistical tests such as:
- T-Test
- ANOVA
  ### Example
Before performing a T-Test on employee salaries, a company wants to check whether the salary data is approximately normally distributed.


```python
from scipy.stats import shapiro

salary = [
    30000, 32000, 31000, 35000,
    34000, 33000, 36000, 31000
]

stat, p_value = shapiro(salary)

print("Shapiro-Wilk Statistic:", stat)
print("P-Value:", p_value)
```
