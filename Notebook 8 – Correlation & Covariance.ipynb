# Correlation & Covariance

## Correlation
Correlation is a statistical method used to measure the relationship between two variables. It tells us whether two variables move together and how strongly they are related.
### Example:
Suppose we compare hours studied and exam marks:
 | Study Hours | Marks |
|---:|---:|
| 1 | 40 |
| 2 | 50 |
| 3 | 60 |
| 4 | 70 |
| 5 | 80 |

```python
import numpy as np

study_hours = np.array([1, 2, 3, 4, 5])
marks = np.array([40, 50, 60, 70, 80])

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Correlation:", correlation)
```

## Covariance
Covariance is a statistical measure that tells us how two variables change together.

In simple words:

- If both variables increase together → Positive covariance
- If one increases while the other decreases → Negative covariance
- If there is no clear relationship → Covariance is close to 0
### Example
Suppose:
- Temperature increases.
- Ice cream sales also increase.

```python
temperature = np.array([20, 25, 30, 35, 40])
sales = np.array([100, 120, 150, 180, 210])

covariance = np.cov(temperature, sales)[0, 1]

print("Covariance:", covariance)
```

## Positive Correlation
Positive correlation means that when one variable increases, the other variable also tends to increase.

Similarly, when one variable decreases, the other variable also tends to decrease.

In simple words:

Both variables move in the same direction.

The correlation value for positive correlation is between 0 and +1.

- +1 → Perfect positive correlation
- +0.8 → Strong positive correlation
- +0.5 → Moderate positive correlation
- +0.1 → Weak positive correlation
  ### Example
Product price increases → Number of customers decreases
| Price | Customers |
|---:|---:|
| 10 | 100 |
| 20 | 80 |
| 30 | 60 |
| 40 | 40 |

```python
price = np.array([10, 20, 30, 40])
customers = np.array([100, 80, 60, 40])

correlation = np.corrcoef(price, customers)[0, 1]

print("Negative Correlation:", correlation)
```

## No Correlation
No Correlation means there is no clear relationship between two variables.

When one variable changes, the other variable does not show a consistent pattern of increasing or decreasing.

The correlation coefficient (r) is usually close to 0.

- r = +1 → Perfect positive correlation
- r = 0 → No linear correlation
- r = -1 → Perfect negative correlation
### Example
A person's:
- shoe size
- exam marks

```python
shoe_size = np.array([6, 7, 8, 9, 10])
marks = np.array([75, 45, 90, 60, 80])

correlation = np.corrcoef(shoe_size, marks)[0, 1]

print("Correlation:", correlation)
```

## Pearson Correlation
Pearson Correlation is a statistical method used to measure the strength and direction of the linear relationship between two numerical variables.
### Formula


Where:
- Cov(X,Y) = covariance between X and Y
- σX = standard deviation of X
- σY = standard deviation of Y
  ### Example
Suppose we want to check whether study hours and marks have a linear relationship.

```python
from scipy.stats import pearsonr

study_hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 80]

correlation, p_value = pearsonr(study_hours, marks)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

## Spearman Correlation
Spearman Correlation is a statistical method used to measure the strength and direction of the relationship between two variables based on their ranks.
It is useful when:
- Data is ordinal/ranked.
- The relationship is not necessarily linear.
- Data contains outliers.
- The relationship is monotonic.
Its value is also between -1 and +1.
| Spearman (ρ) | Meaning                       |
| ------------ | ----------------------------- |
| **+1**       | Perfect positive relationship |
| **0**        | No monotonic relationship     |
| **-1**       | Perfect negative relationship |
### Example
Suppose students are ranked based on:

- Study hours
- Exam performance

Even if the difference between marks is not equal, if students who study more generally receive higher marks, Spearman correlation can identify this relationship using their ranks.

```python
from scipy.stats import spearmanr

hours = [2, 4, 6, 8, 10]
marks = [45, 55, 70, 80, 95]

correlation, p_value = spearmanr(hours, marks)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

## Correlation Matrix
A Correlation Matrix is a table that shows the correlation between multiple variables in a dataset.

Instead of calculating the correlation between only two variables, a correlation matrix shows the relationship between every pair of variables.
### Example
A dataset may contain:
- Study Hours
- Attendance
- Assignment Marks
- Exam Marks

The correlation matrix shows the relationship between every pair of these variables.
| | Study Hours | Attendance | Exam Marks |
|---|---:|---:|---:|
| Study Hours | 1.00 | 0.70 | 0.85 |
| Attendance | 0.70 | 1.00 | 0.75 |
| Exam Marks | 0.85 | 0.75 | 1.00 |

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 5, 7, 8],
    "Attendance": [60, 70, 75, 85, 90],
    "Exam_Marks": [45, 55, 65, 75, 85]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
```

## Heatmap
Heatmap is a graphical way of representing data using different shades of color.

In correlation analysis, a heatmap is used to visually display the Correlation Matrix.

A correlation heatmap makes it easy to see which variables have:
- Strong positive correlation
- Strong negative correlation
- Weak correlation
A correlation heatmap converts a correlation matrix into an easy-to-understand visual representation.
### Example
Imagine a company has these features:
- Age
- Experience
- Salary
- Working Hours
- Performance Score

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create employee dataset
data = {
    "Age": [22, 25, 28, 32, 35, 40, 45, 50],
    "Experience": [1, 3, 5, 8, 10, 15, 20, 25],
    "Salary": [25000, 32000, 40000, 55000, 65000, 80000, 100000, 120000],
    "Working_Hours": [8, 8, 9, 8, 9, 10, 9, 8],
    "Performance_Score": [60, 65, 70, 78, 82, 88, 92, 95]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate correlation matrix
correlation_matrix = df.corr()

print("Correlation Matrix:")
print(correlation_matrix)

# Create heatmap
plt.figure(figsize=(9, 6))

plt.imshow(correlation_matrix, interpolation="nearest")

# Add labels
plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

# Display correlation values
for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix.columns)):
        plt.text(
            j,
            i,
            round(correlation_matrix.iloc[i, j], 2),
            ha="center",
            va="center"
        )

plt.colorbar(label="Correlation")
plt.title("Employee Correlation Heatmap")
plt.tight_layout()
plt.show()
```
