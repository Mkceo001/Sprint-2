# Data Distribution

## Normal Distribution
Understand the Concept
A Normal Distribution is a common way in which numerical data is distributed around its average value.
It has a bell-shaped curve. Most values are close to the mean, while fewer values occur farther away from the mean.
In a perfectly normal distribution:
- Mean = Median = Mode
- The curve is symmetric.
- Most observations are near the center.
- Very large and very small values are less common.
Important terms
- Mean: Center of the distribution.
- Standard deviation: Measures how spread out the data is.
- Bell curve: Shape of a normal distribution.
### Example
if the heights of a large group of people are normally distributed, most people will have heights close to the average, while very short and very tall people will be less common.
### AI/ML Example
Many machine learning algorithms work with numerical features where understanding their distribution is useful.
#### For example:
- Age
- Height
- Measurement errors
- Sensor readings
### Numerical Example
Suppose the average exam score is 70 and the standard deviation is 10.
The scores might look like:
40  50  55  60  65  68  70  72  75  80  85  90  100

                  = 70 Mean

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample marks of 100 students
marks = np.random.normal(loc=70, scale=10, size=100)

plt.hist(marks, bins=10, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

## Uniform Distribution
Uniform Distribution is a type of distribution where all possible values have an equal or approximately equal chance of occurring
### Example
when rolling a fair six-sided die, each number from 1 to 6 has the same probability.
1   2   3   4   5   6


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random values between 1 and 7
# Each interval [1,2), [2,3), ... [6,7) has equal probability
random_values = np.random.uniform(1, 7, 1000)

# Convert continuous values into die numbers 1 to 6
die_rolls = np.floor(random_values).astype(int)

# Display first 20 rolls
print(die_rolls[:20])

# Plot the distribution
plt.hist(die_rolls, bins=np.arange(1, 8) - 0.5, edgecolor="black")
plt.xlabel("Die Number")
plt.ylabel("Frequency")
plt.title("Uniform Distribution - Fair Six-Sided Die")
plt.xticks(range(1, 7))
plt.show()
```

## Gaussian Distribution
Gaussian Distribution is another name for the Normal Distribution.

It is a probability distribution where most values are concentrated around the mean (average), and fewer values occur as we move farther away from the mean.
It is described using two main parameters:
- Mean (μ) – determines the center.
- Standard deviation (σ) – determines the spread.
  

### Example 
Mean = 50
Standard Deviation = 5


```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 5

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30)
plt.title("Gaussian Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## Left Skew
Left Skew is a type of data distribution where the tail of the distribution extends toward the left side, while most of the data values are concentrated on the right side.

It is also called Negative Skewness.

\[
Mean < Median < Mode
\]
### Example
magine an area where most houses cost around:

₹40 lakh – ₹60 lakh

but a few houses are unusually cheap:

₹10 lakh
₹15 lakh

Those low values create a long tail on the left.


```python
import numpy as np
import matplotlib.pyplot as plt

# House prices in lakhs
house_prices = [
    10, 15, 20,
    40, 42, 45, 45, 48, 50,
    50, 52, 55, 55, 58, 60,
    60, 62, 65, 68
]

# Calculate mean, median and mode
mean = np.mean(house_prices)
median = np.median(house_prices)

print("Mean:", mean)
print("Median:", median)

# Create histogram
plt.hist(house_prices, bins=8, edgecolor="black")

plt.xlabel("House Price (₹ Lakhs)")
plt.ylabel("Number of Houses")
plt.title("Left-Skewed Distribution of House Prices")

plt.show()
```

## Right Skew
Right Skew is a type of data distribution where the tail extends toward the right side, while most of the data values are concentrated on the left side.

It is also called Positive Skewness.

\(Mode < Median < Mean\)
### Example: 

Imagine the salaries of employees in a company:

₹20,000,
₹22,000,
₹25,000,
₹25,000,
₹28,000,
₹30,000,
₹30,000,
₹32,000,
₹35,000,
₹1,50,000.

Most employees earn around ₹20,000–₹35,000, but a few employees earn a very high salary such as ₹1,50,000.

Mode   = ₹25,000,
Median = ₹30,000,
Mean   = ₹42,000

```python
import numpy as np
import matplotlib.pyplot as plt

# Employee salaries in ₹
salaries = [
    20000, 22000, 25000, 25000, 28000,
    30000, 30000, 32000, 35000,
    150000
]

# Calculate mean and median
mean = np.mean(salaries)
median = np.median(salaries)

print("Mean:", mean)
print("Median:", median)

# Create histogram
plt.hist(salaries, bins=8, edgecolor="black")

plt.xlabel("Salary (₹)")
plt.ylabel("Number of Employees")
plt.title("Right-Skewed Distribution of Employee Salaries")

plt.show()
```

## Symmetric Distribution
A Symmetric Distribution is a distribution where the data is evenly balanced on both sides of the center.
If the left side of a distribution looks like a mirror image of the right side, the distribution is symmetric.
\(Mean = Median = Mode\)
### Example:
Suppose students' marks are distributed like this:

Marks:       40  50  60  70  80  90  100
Students:     2   5  10  15  10   5    2

The number of students is balanced around 70:

          2   5   10   15   10   5   2
          ↓   ↓    ↓    ↓    ↓   ↓   ↓
         40  50   60   70   80  90  100
                      ↑
                    Center

Notice:

40 and 100 → 2 students
50 and 90 → 5 students
60 and 80 → 10 students
70 → 15 students

The two sides are balanced.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate symmetric data
data = np.random.normal(70, 10, 1000)

# Create histogram
plt.hist(data, bins=30, edgecolor="black")

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")

plt.show()
```

## Kurtosis
Kurtosis is a statistical measure that describes the shape of a distribution, especially how much data is in the tails compared with a normal distribution.
here are three commonly discussed types:
1. Mesokurtic
Similar tail behavior to a normal distribution.- K = 0
2. Leptokurtic
Has heavier tails, meaning extreme values are more common.- K > 0
3. Platykurtic
Has lighter tails, meaning extreme values are less common.- K< 0

Leptokurtic  → heavier tails

Mesokurtic   → normal-like

Platykurtic  → lighter tails


```python
import numpy as np
from scipy.stats import kurtosis

data = np.random.normal(0, 1, 1000)

result = kurtosis(data)

print("Kurtosis:", result)
```

## Empirical Rule – 68–95–99.7
The Empirical Rule, also called the 68–95–99.7 Rule, tells us how data is distributed around the mean in a normal (bell-shaped) distribution.

In a normal distribution, about 68% of the data lies within 1 standard deviation, 95% within 2 standard deviations, and 99.7% within 3 standard deviations of the mean.


```python
import numpy as np

# Generate normally distributed data
data = np.random.normal(100, 10, 100000)

mean = np.mean(data)
std = np.std(data)

within_1 = np.mean((data >= mean - std) & (data <= mean + std)) * 100
within_2 = np.mean((data >= mean - 2 * std) & (data <= mean + 2 * std)) * 100
within_3 = np.mean((data >= mean - 3 * std) & (data <= mean + 3 * std)) * 100

print("Mean:", mean)
print("Standard Deviation:", std)
print("Within 1 SD:", within_1, "%")
print("Within 2 SD:", within_2, "%")
print("Within 3 SD:", within_3, "%")
```
