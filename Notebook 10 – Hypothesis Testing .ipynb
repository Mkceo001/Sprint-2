# Hypothesis Testing 

Hypothesis Testing is a statistical method used to make a decision or conclusion about a population using sample data.

In simple words, we start with an assumption and use data to check whether there is enough evidence to support or reject that assumption.
### Example
A company claims:
"Our new training program increases employee productivity."
##### Basic Process:

Collect Sample Data
       ↓
Create Hypotheses
       ↓
Choose Significance Level
       ↓
Calculate Test Statistic / p-value
       ↓
Make Decision
       ↓
Accept or Reject the Evidence Against H₀

##### Important Terminology

| Term | Meaning |
|---|---|
| Null Hypothesis | Default assumption |
| Alternative Hypothesis | Claim we want to find evidence for |
| p-value | Evidence against the null hypothesis |
| Significance Level | Threshold used for decision-making |
| Confidence Interval | Range of plausible values |
| Type I Error | Rejecting a true null hypothesis |
| Type II Error | Failing to reject a false null hypothesis |

##### Two Main Hypotheses
1. Null Hypothesis (H₀)

The null hypothesis represents the default assumption or "no change/no difference" situation.

##### Example:

H₀: The average delivery time is 30 minutes.

2. Alternative Hypothesis (H₁ or Hₐ)

The alternative hypothesis represents what we want to check or prove.

##### Example:

H₁: The average delivery time is not 30 minutes.

## Null Hypothesis
The Null Hypothesis (H₀) is the default assumption in hypothesis testing.

It usually says that there is no significant difference, no significant relationship, or no effect in the situation we are testing.

It usually states that:
- There is no difference.
- There is no effect.
- There is no relationship.
For example:
A company wants to know whether a new training program changes employee productivity.

The null hypothesis could be:



This means there is no difference in average productivity.

```python
from scipy import stats

sample = [52, 55, 51, 53, 56, 54, 52, 55, 57, 53]

t_stat, p_value = stats.ttest_1samp(sample, 50)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

## Alternative Hypothesis
The Alternative Hypothesis is the statement that there is a significant difference, relationship, or effect in the population.
##### Types of Alternative Hypothesis

There are mainly three forms.

1. Two-Tailed Alternative Hypothesis

Checks whether the value is different in either direction.

H₁: μ ≠ 50

Example:

The average delivery time is different from 30 minutes.

2. Right-Tailed Alternative Hypothesis

Checks whether the value is greater than a specific value.

H₁: μ > 50

Example:

The new ML model has accuracy greater than 90%.

3. Left-Tailed Alternative Hypothesis

Checks whether the value is less than a specific value.

H₁: μ < 50

Example:

The new algorithm reduces average processing time below 50 milliseconds.
### Example
Suppose a website team introduces a new design.
They want to know:
"Does the new design change the average time users spend on the website?"

H₀: New design has no effect.

H₁: New design has an effect.

```python
from scipy import stats

old_average = 50

sample = [52, 55, 51, 53, 56, 54, 52, 55, 57, 53]

t_stat, p_value = stats.ttest_1samp(sample, old_average)

print("P-value:", p_value)
```

## p-value
The p-value (probability value) is a number used in hypothesis testing to help us decide whether the evidence against the Null Hypothesis (H₀) is strong enough.
##### Common Rule

We usually choose a significance level (α) = 0.05.
| P-value      | Decision          | Meaning                         |
| ------------ | ----------------- | ------------------------------- |
| **p < 0.05** | Reject H₀         | Significant evidence against H₀ |
| **p ≥ 0.05** | Fail to reject H₀ | Not enough evidence against H₀  |
### Example

```python
from scipy import stats

sample = [52, 55, 51, 53, 56, 54, 52, 55, 57, 53]

t_stat, p_value = stats.ttest_1samp(sample, 50)

print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## Confidence Interval
A Confidence Interval (CI) is a range of values that is used to estimate an unknown population parameter, such as the population mean.
#### Confidence Level
| Confidence Level | Meaning                 |
| ---------------- | ----------------------- |
| **90%**          | 90% confidence interval |
| **95%**          | Most commonly used      |
| **99%**          | Higher confidence       |
#### Basic Formula

For a population mean, a simplified formula is:

$$ CI = \bar{x} \pm Margin\ of\ Error $$

Where:

- \(\bar{x}\) = Sample mean
- Margin of Error = amount added and subtracted from the sample mean

For a large sample with known population standard deviation:

$$ CI = \bar{x} \pm Z \frac{\sigma}{\sqrt{n}} $$

Where:

- \(Z\) = critical value
- \(\sigma\) = population standard deviation
- \(n\) = sample size
### Example

Suppose an AI model is tested on 100 datasets.

The average accuracy is:

Mean accuracy = 90%

After calculation:

95% Confidence Interval = 88% to 92%

So we report:

The estimated average accuracy is 90%, with a 95% confidence interval of 88%–92%.

This means the data gives us a plausible range of 88% to 92% for the population's true average accuracy.

```python
import numpy as np
from scipy import stats

# Sample data
salary = np.array([
    28000, 30000, 32000, 29000, 31000,
    33000, 30000, 29000, 31000, 32000
])

# Sample mean
mean = np.mean(salary)

# Calculate 95% confidence interval
confidence_interval = stats.t.interval(
    confidence=0.95,
    df=len(salary) - 1,
    loc=mean,
    scale=stats.sem(salary)
)

print("Sample Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```

## Significance Level
The Significance Level, represented by α (alpha), is a value we choose before performing a hypothesis test.

It tells us the maximum probability of making a Type I error that we are willing to accept.

The most common significance level is:

α = 0.05

In simple words:

Significance level is the cutoff we use to decide whether a result is statistically significant.
### Example
Suppose:
α = 0.05
p-value = 0.03

```python
alpha = 0.05
p_value = 0.03

if p_value < alpha:
    print("Statistically significant")
else:
    print("Not statistically significant")
```

## Type I Error
A Type I Error occurs when we reject the Null Hypothesis (H₀) even though it is actually true.

In simple words:

Type I Error = False Positive

We conclude that an effect or difference exists, but in reality, there is no effect or difference
### Example

Imagine a medical test for a disease.

H₀: The person does not have the disease.
H₁: The person has the disease.

If the test says:

"The person has the disease."

but the person actually does not have the disease, this is a Type I Error.

It is called a false positive.


### Type II Error
A Type II Error occurs when we fail to reject the Null Hypothesis (H₀) even though it is actually false.

In simple words:

Type II Error = False Negative

We conclude that there is not enough evidence of an effect, but in reality, an effect or difference does exist.
##### Type I vs Type II Error
|                       | H₀ is Actually True | H₀ is Actually False |
| --------------------- | ------------------- | -------------------- |
| **Reject H₀**         | ❌ Type I Error      | ✅ Correct            |
| **Fail to Reject H₀** | ✅ Correct           | ❌ Type II Error      |

##### Key Point

Type II Error = Failing to reject a false Null Hypothesis.

Type I → False Positive → α

Type II → False Negative → β

### Example

Imagine a medical test.

H₀: The person does not have the disease.
H₁: The person has the disease.

Suppose the person actually has the disease, but the test says:

"There is not enough evidence that the person has the disease."

This is a Type II Error.

It is called a false negative.

## One-Tailed Test
A One-Tailed Test is a hypothesis test used when we are interested in finding out whether a population parameter is greater than or less than a specific value.
##### There are two types:

- Right-Tailed Test → checks whether the value is greater
- Left-Tailed Test → checks whether the value is less
#### Right-Tailed Test

A right-tailed test checks whether the value is greater than a specific value.

##### Example

Suppose an AI model currently has 90% accuracy.

We want to know whether a new model has higher accuracy.
#### Left-Tailed Test

A left-tailed test checks whether the value is less than a specific value.

##### Example

Suppose an algorithm takes an average of 100 milliseconds to process an image.

We want to know whether a new algorithm is faster, meaning it takes less than 100 ms.
### Example
Suppose a company claims:
"The new machine produces more than 100 products per hour."

We test:
H₀: μ ≤ 100

H₁: μ > 100

```python
from scipy import stats

data = [105, 110, 108, 112, 107, 111, 109, 106, 113, 110]

t_stat, p_two = stats.ttest_1samp(data, 100)

if t_stat > 0:
    p_one = p_two / 2
else:
    p_one = 1 - (p_two / 2)

print("T-statistic:", t_stat)
print("One-tailed p-value:", p_one)
```

## Two-Tailed Test  
A Two-Tailed Test is a hypothesis test used when we want to know whether a population value is different from a specific value.

It checks for differences in both directions:

The value could be greater than the expected value.
The value could be less than the expected value.

In simple words:

A two-tailed test checks whether there is a significant difference, without specifying the direction of the difference.
### example:
Does the new training program change productivity?

We don't specify whether productivity will increase or decrease.

The hypotheses are:

Null Hypothesis (H₀):

$$ H_0: \mu = 50 $$

Alternative Hypothesis (H₁):



The alternative hypothesis allows both:


```python
from scipy import stats

data = [105, 110, 108, 112, 107, 111, 109, 106, 113, 110]

t_stat, p_value = stats.ttest_1samp(data, 100)

print("T-statistic:", t_stat)
print("Two-tailed p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

#### One-Tailed vs Two-Tailed
| One-Tailed Test             | Two-Tailed Test                  |
| --------------------------- | -------------------------------- |
| Checks one direction        | Checks both directions           |
| `>` or `<`                  | `≠`                              |
| Example: μ > 100            | Example: μ ≠ 100                 |
| One rejection region        | Two rejection regions            |
| Used when direction matters | Used when any difference matters |
