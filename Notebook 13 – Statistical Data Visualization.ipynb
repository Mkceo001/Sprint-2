# Statistical Data Visualization
Statistical data visualization helps us understand data using charts and graphs. In AI/ML, visualization is mainly used during Exploratory Data Analysis (EDA) to find distributions, relationships, trends, outliers, and patterns.


## Histogram
A Histogram shows the distribution of numerical data by dividing values into intervals called bins.

#### Example: 
Distribution of employee salaries
#### Purpose
A Histogram shows the distribution of numerical data by grouping values into intervals called bins.
It helps us understand:
- Data distribution
- Center
- Spread
- Skewness
- Possible outliers
When to Use
#### Use a histogram when:
- You have numerical data.
- You want to understand its distribution.
- You want to check whether data is approximately normal.
- You want to identify skewness.
#### Advantages
- Easy to understand.
- Shows the shape of the distribution.
- Helps identify skewness and unusual values.
- Useful for EDA.

```python
import matplotlib.pyplot as plt

age = [22, 25, 28, 30, 32, 35, 35, 36, 38, 40, 42, 45, 50]

plt.hist(age, bins=5)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

## Bar Chart
A Bar Chart is a statistical visualization used to compare values between different categories.

Each category is represented by a bar, and the length or height of the bar shows its value.
#### Use a bar chart when:
- Comparing categories.
- Working with counts or aggregated values.
- Comparing sales, departments, products, etc.
#### Advantages
- Very easy to understand.
- Excellent for category comparison.
- Can show large differences clearly.
  ### Example

Suppose we have the number of employees in different departments:
| Department | Employees |
| ---------- | --------: |
| IT         |        50 |
| HR         |        30 |
| Finance    |        40 |
| Sales      |        60 |



```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 250, 100, 80]

plt.bar(products, sales)

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

## A Line Chart is a statistical visualization used to show changes or trends in data over time or across an ordered sequence.

Points are plotted on the chart and connected using lines. This makes it easy to see whether values are increasing, decreasing, or staying stable.
#### Use a line chart for:
- Time-series data.
- Monthly sales.
- Daily temperature.
- Website traffic.
- Stock prices.
#### Advantages
- Clearly shows trends.
- Good for time-series data.
- Makes increases and decreases easy to identify.
### Example

Suppose a company's monthly sales are:

| Month    | Sales |
| -------- | ----: |
| January  |   100 |
| February |   120 |
| March    |   150 |
| April    |   140 |
| May      |   180 |


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 115, 150, 170, 190]

plt.plot(months, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")

plt.show()
```

## Scatter Plot
A Scatter Plot is a statistical visualization used to show the relationship between two numerical variables.

It represents each observation as a point (dot) on a graph.
#### Use a scatter plot when:
- Both variables are numerical.
- You want to identify relationships.
- You want to identify clusters or outliers.
#### Advantages
- Clearly shows relationships.
- Helps identify positive/negative correlation.
- Can reveal clusters.
- Can reveal outliers.
### Example
we can study the relationship between:

Experience and Salary
| Experience | Salary |
| ---------: | -----: |
|          1 | 25,000 |
|          3 | 32,000 |
|          5 | 40,000 |
|          8 | 55,000 |
|         10 | 65,000 |


```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 45, 52, 65, 75, 85]

plt.scatter(hours, marks)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.title("Study Hours vs Exam Marks")

plt.show()

```

## Pie Chart
A Pie Chart is a statistical visualization used to show how a whole is divided into different categories.

The complete circle represents 100%, and each slice represents one category's share of the total.
#### Use it when:
- There are only a few categories.
- You want to show percentages or proportions.
- All categories together represent 100%.
#### Advantages
- Easy to understand for a small number of categories.
- Good for showing proportions.
- Useful for simple business presentations.
  ### Example

Suppose a company's employees are divided into departments
| Department | Employees |
| ---------- | --------: |
| IT         |        50 |
| HR         |        20 |
| Finance    |        30 |



```python
import matplotlib.pyplot as plt

departments = ["IT", "HR", "Finance"]
employees = [50, 20, 30]

plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%"
)

plt.title("Employee Distribution by Department")

plt.show()
```

## Density Plot
A Density Plot is a statistical visualization used to show the distribution of numerical data.

It displays a smooth curve that represents where the data values are concentrated.
#### Use it when:
- You want to understand the shape of numerical data.
- You want to compare distributions.
- You want to identify peaks and skewness.
#### Advantages
- Smooth representation.
- Easy to compare multiple distributions.
- Shows peaks and distribution shape clearly
### Example

Suppose we have students' exam marks:

45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 80

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample data
data = np.random.normal(60, 10, 500)

# Create density plot
sns.kdeplot(data=data)

plt.xlabel("Marks")
plt.ylabel("Density")
plt.title("Density Plot of Student Marks")

plt.show()
```

## Box plot
A Box Plot (also called a Box-and-Whisker Plot) is a statistical visualization used to understand the distribution, spread, median, and outliers of numerical data.
##### Box Plot summarizes the distribution of numerical data using:
- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum
  #### Use a box plot when:
- You want to detect outliers.
- You want to compare distributions.
- You want to understand spread and median.
#### Advantages
- Excellent for outlier detection.
- Compact representation.
- Easy to compare multiple groups.
- Shows median and quartiles.    

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 95]

plt.boxplot(marks)

plt.ylabel("Marks")
plt.title("Box Plot of Student Marks")

plt.show()
```

## Violin Plot
A Violin Plot combines features of a Box Plot and Density Plot.
It shows:
- Median
- Spread
- Distribution shape
- Density
##### When to Use
Use a violin plot when:
- Comparing distributions across groups.
- You want more information than a box plot provides.
- You want to see where observations are concentrated.
##### Advantages
- Shows distribution shape.
- Shows density.
- Good for comparing multiple groups.
- Provides more information than a simple box plot.
  

```python
import matplotlib.pyplot as plt

group_a = [20, 22, 23, 25, 26, 28, 30]
group_b = [30, 32, 35, 36, 38, 40, 42]

plt.violinplot([group_a, group_b])

plt.xticks([1, 2], ["Group A", "Group B"])
plt.ylabel("Score")
plt.title("Score Distribution")

plt.show()
```

## Pair Plot
A Pair Plot is a statistical visualization used to show the relationships between multiple numerical variables at the same time.
#### Use a pair plot during:
- Exploratory Data Analysis.
- Feature relationship analysis.
- Initial dataset investigation.
#### Advantages
- Shows many relationships at once.
- Useful for identifying correlations.
- Can reveal clusters.
- Can reveal outliers.
### Example
if a dataset has:

- Age
- Experience
- Salary
  

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Age": [22, 25, 28, 32, 35, 40],
    "Experience": [1, 3, 5, 8, 10, 15],
    "Salary": [25000, 32000, 40000, 55000, 65000, 80000]
}

df = pd.DataFrame(data)

sns.pairplot(df)

plt.show()
```

## Heatmap
A Heatmap is a statistical visualization that represents numerical values using different levels of shading or color intensity.
#### Use a heatmap when:
- You have a matrix of values.
- You want to visualize correlations.
- You want to identify strong relationships quickly.
#### Advantages
- Easy to identify patterns.
- Excellent for correlation matrices.
- Can display many relationships in a compact form.
### Example
|            |  Age | Experience | Salary |
| ---------- | ---: | ---------: | -----: |
| Age        | 1.00 |       0.95 |   0.90 |
| Experience | 0.95 |       1.00 |   0.98 |
| Salary     | 0.90 |       0.98 |   1.00 |


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Age": [22, 25, 28, 32, 35, 40, 45, 50],
    "Experience": [1, 3, 5, 8, 10, 15, 20, 25],
    "Salary": [25000, 32000, 40000, 55000, 65000, 80000, 100000, 120000]
}

df = pd.DataFrame(data)

# Calculate correlation
correlation = df.corr()

# Create heatmap
sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```
