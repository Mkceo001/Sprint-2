## Why Statistics is Important in AI/ML ?

Machine learning works with data. Statistics gives us tools to understand that data before, during, and after model development.

Statistics helps an AI/ML engineer:
- Understand datasets
- Find patterns
- Detect unusual values
- Measure relationships between variables
- Handle missing values
- Evaluate models
- Estimate uncertainty
- Test hypotheses
- Understand model performance

### Example :
Imagine a company wants to predict whether customers will purchase a product.
| Age | Income | Previous Purchases | Purchased |
|---:|---:|---:|---|
| 22 | 25000 | 1 | No |
| 28 | 35000 | 2 | No |
| 35 | 50000 | 5 | Yes |
| 42 | 70000 | 8 | Yes |

```python
import pandas as pd

data = {
    "Age": [22, 28, 35, 42],
    "Income": [25000, 35000, 50000, 70000],
    "Previous_Purchases": [1, 2, 5, 8],
    "Purchased": ["No", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)

print("\nAverage Age:", df["Age"].mean())
print("Average Income:", df["Income"].mean())
```

## Statistics vs Machine Learning

Statistics and machine learning both work with data, but their primary goals can be different.
### Statistics
Statistics often focuses on:
Understanding data and drawing conclusions from it.

Statistics → Understand

Machine Learning → Predict

#### Example:
- What is the average?
- Is there a relationship between two variables?
- Is this difference statistically significant?
- What can we conclude about a population?
### Machine Learning
Machine learning often focuses on:
Learning patterns from data to make predictions or decisions.

#### Example:
- Will this customer purchase?
- What will tomorrow's sales be?
- Is this image a cat or dog?

They overlap heavily. Statistical concepts are used throughout machine learning.

```python
import numpy as np

house_size = np.array([500, 800, 1000, 1200, 1500])
house_price = np.array([20, 30, 40, 48, 60])

correlation = np.corrcoef(house_size, house_price)[0, 1]

print("Correlation:", correlation)
```

## Descriptive Statistics

Descriptive statistics is the branch of statistics used to organize, summarize, and describe data in a clear and meaningful way
    
Common descriptive statistics include:
- Mean
- Median
- Mode
- Minimum
- Maximum
- Range
- Variance
- Standard deviation
- Percentiles  
### Example :
instead of displaying 10,000 employee salaries individually, we can summarize them using:
Average salary = ₹45,000

Minimum salary = ₹20,000

Maximum salary = ₹90,000

The program calculates several descriptive statistics

A higher standard deviation means the values are more spread out around the mean.

Descriptive statistics are commonly used during Exploratory Data Analysis (EDA) before model training.

```python
import numpy as np

sales = np.array([100, 120, 150, 130, 200])

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Range:", np.max(sales) - np.min(sales))
print("Standard Deviation:", np.std(sales))
```

## Inferential Statistics

Inferential statistics is a branch of statistics used to make conclusions, predictions, or decisions about a larger population using data collected from a smaller sample.
We can use statistical methods to estimate or test claims about the larger student population.

Important concepts include:
- Sampling
- Confidence intervals
- Hypothesis testing
- Statistical significance
- Probability distributions
 ### Example:

```python
import numpy as np

np.random.seed(42)

population = np.random.normal(70, 10, 10000)

sample = np.random.choice(population, size=100)

print("Population mean:", np.mean(population))
print("Sample mean:", np.mean(sample))
```

## Sample
A sample is a smaller group selected from a larger population for the purpose of collecting and analyzing data.
### Example:
Suppose an e-commerce company has one million customers.
Studying every customer may be expensive.
The company randomly selects 5,000 customers and studies their purchasing behavior.
1,000,000 customers → Population
5,000 customers → Sample


```python
import numpy as np

population = np.arange(1, 1001)

sample = np.random.choice(population, size=50, replace=False)

print("Population size:", len(population))
print("Sample size:", len(sample))
print("Sample:", sample)
```

## Cencus
A census is a method of collecting data from every member of the entire population being studied.
### Example
If a company has 10,000 employees and collects salary information from all 10,000 employees, it is conducting a census.

Sample → Some members

Census → Every member

```python
import numpy as np

students = np.array([60, 65, 70, 75, 80, 85, 90])

# Census: use every student
census_mean = np.mean(students)

# Sample: use only selected students
sample = students[[0, 2, 4, 6]]
sample_mean = np.mean(sample)

print("Census mean:", census_mean)
print("Sample mean:", sample_mean)
```

## Sampling

Sampling is the process of selecting a smaller group (sample) from a larger population to collect and analyze data.
Common sampling methods include:
- Simple random sampling
- Systematic sampling
- Stratified sampling
- Cluster sampling
### Simple Example

Suppose a college has 5,000 students and you want to study their study habits.

Population: 5,000 students
Sampling: Selecting 500 students from the 5,000
Sample: The 500 selected student

```python
import numpy as np

np.random.seed(42)

students = np.arange(1, 101)

sample = np.random.choice(
    students,
    size=10,
    replace=False
)

print("Selected sample:")
print(sample)
```

## Features
A feature is an individual measurable property or characteristic of data that is used to describe an observation and, often, to help make predictions.
### Example:
Suppose we want to predict the price of a house:
| House | Area (sq.ft) | Bedrooms | Age (years) | Location | Price |
| ----- | -----------: | -------: | ----------: | -------- | ----: |
| A     |        1,000 |        2 |          10 | Chennai  | ₹50 L |
| B     |        1,500 |        3 |           5 | Chennai  | ₹75 L |
| C     |        2,000 |        4 |           2 | Chennai  | ₹1 Cr |
- Area → Feature
- Number of bedrooms → Feature
- House age → Feature
-Location → Feature
Price → Target/Label (what we want to predict)


```python
import pandas as pd

data = {
    "Area": [1000, 1500, 2000],
    "Bedrooms": [2, 3, 4],
    "Age": [10, 5, 3],
    "Price": [40, 60, 80]
}

df = pd.DataFrame(data)

features = df[["Area", "Bedrooms", "Age"]]
target = df["Price"]

print("Features:")
print(features)

print("\nTarget:")
print(target)
```

## Unstructured Data
Unstructured data is data that does not follow a fixed or predefined format or structure. it is data that is not organized neatly into rows and columns.

### Examples
- Images – photos, medical scans
- Videos – YouTube videos, CCTV footage
- Audio – voice recordings, songs, podcasts
- Text documents – emails, Word documents, PDF files
- Social media posts – comments, posts, messages
- Customer reviews – written feedback
### Example
A customer writes:

"I really like this phone. The camera quality is excellent, but the battery could be better."

This is unstructured text data because it does not follow a fixed table format.



```python
import numpy as np

# Simulate a small grayscale image
image = np.array([
    [0, 255, 100],
    [50, 200, 150],
    [255, 0, 75]
])

print("Image:")
print(image)

print("\nImage shape:", image.shape)
print("Average pixel value:", np.mean(image))
```
