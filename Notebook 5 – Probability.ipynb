The value of probability is always between 0 and 1.

0 → Impossible to happen

1 → Certain to happen

Between 0 and 1 → Possible to happen

## Events
In probability, an event is a specific outcome or a group of outcomes that we are interested in from a random experiment.
### Exmple 
when rolling a die we might be interested in getting an even number.
The possible outcomes are:
1, 2, 3, 4, 5, 6

The event "getting an even number" is:
2, 4, 6
### Types of Events

There are several important types of events in probability.

##### 1. Simple Event

A simple event contains only one outcome.

Example:

Rolling a dice and getting 5.

Event = {5}
##### 2. Compound Event

A compound event contains more than one outcome.

Example:

Rolling a dice and getting an even number.

Event = {2, 4, 6}

There are three possible favorable outcomes.

##### 3. Certain Event

An event that will definitely happen.

Example:

Rolling a standard dice and getting a number between 1 and 6.

{1, 2, 3, 4, 5, 6}

Every possible outcome satisfies the condition.

Therefore:

$$ P(E) = 1 $$
##### 4. Impossible Event

An event that cannot happen.

Example:

Rolling a standard dice and getting 10.

A standard dice only contains:

1, 2, 3, 4, 5, 6

So:

$$ P(E) = 0 $$
##### 5. Complementary Event

The complement of an event means the event does not happen.

Example:

If:

Event A = Getting Heads

Then:

Complement of A = Getting Tails

```python
outcomes = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in outcomes if x % 2 == 0]

probability = len(even_numbers) / len(outcomes)

print("Even number outcomes:", even_numbers)
print("Probability:", probability)
```

## Sample Space
The sample space is the set of all possible outcomes of a random experiment.
### Example:
Suppose we toss two coins.

The possible outcomes are:

- Heads - Heads
- Heads - Tails
- Tails - Heads
- Tails - Tails

So the sample space is:

$$ S = \{HH, HT, TH, TT\} $$

There are 4 possible outcomes.

If the event is:

Getting two Heads

Then:

$$ E = \{HH\} $$

Therefore:

$$ P(E) = \frac{1}{4} = 25\% $$

```python
sample_space = ["HH", "HT", "TH", "TT"]

print("Sample Space:")
for outcome in sample_space:
    print(outcome)

print("Total outcomes:", len(sample_space))
```

## Independent Events
Two events are independent when the occurrence of one event does not affect the probability of the other event.
### Example:
- Tossing a coin
- Rolling a die
The result of the coin toss does not change the result of the die roll.
### Formula

If A and B are independent events:

$$ P(A \cap B) = P(A) \times P(B) $$

Here:

- \(P(A)\) = Probability of event A
- \(P(B)\) = Probability of event B
- \(P(A \cap B)\) = Probability that A and B both happen
### Real-Life Example

Imagine two different students independently answer questions in an online quiz.

Whether Student A answers correctly does not affect whether Student B answers correctly.



```python
# Probability of two independent events

# Probability that Student A answers correctly
student_a = 0.8

# Probability that Student B answers correctly
student_b = 0.7

# Since the events are independent,
# multiply their probabilities
probability_both_correct = student_a * student_b

print("Probability of Student A answering correctly:", student_a)
print("Probability of Student B answering correctly:", student_b)
print("Probability that both answer correctly:", probability_both_correct)
print("Probability in percentage:", probability_both_correct * 100, "%")
```

## Dependent Events
Two events are dependent when the result of one event affects the probability of another event.
A common example is drawing cards from a deck without replacing the first card.
After one card is removed, the number of remaining cards changes.
### Example:
- 3 red balls
- 2 blue balls
  
Tatal 5 balls
Probability of drawing a red ball first.
If we do not replace that red ball, only 4 balls remain.

Now
- 2 red
- 2 blue
  

```python
prob_first_red = 3 / 5
prob_second_red = 2 / 4

prob_both_red = prob_first_red * prob_second_red

print("Probability of first red:", prob_first_red)
print("Probability of second red:", prob_second_red)
print("Probability of two red balls:", prob_both_red)
```

## Easy way to remember:

Independent = No effect

Dependent = Previous event has an effect

## Conditional Probability
Conditional probability tells us the probability of an event when we already know that another event has happened.
Formula

The formula for conditional probability is:

$$ \boxed{P(A|B)=\frac{P(A\cap B)}{P(B)}} $$

Where:

- P(A|B) → Probability of A given B
- P(A ∩ B) → Probability that both A and B happen
- P(B) → Probability of B happening
  ### Example:
Suppose a company has 100 customers:
- 60 use a mobile app.
- 30 customers both use the app and purchase a product.
What is the probability that a customer purchased the product given that they use the app?

So the probability is 50%.

```python
app_users = 60
app_users_who_purchased = 30

conditional_probability = app_users_who_purchased / app_users

print("P(Purchase | App User):", conditional_probability)
```

## Bayes' Theorem
### What is Bayes' Theorem?

Bayes' Theorem is a mathematical method used to calculate the probability of an event when we have some new information or evidence.
Bayes' Theorem Formula
$$ \boxed{ P(A|B)=\frac{P(B|A)\times P(A)}{P(B)} } $$

Where:

- \(P(A|B)\) → Probability of A given B
- \(P(B|A)\) → Probability of B given A
- \(P(A)\) → Prior probability of A
- \(P(B)\) → Probability of the evidence B
#### Example
  Suppose there is a disease called Disease X.

Out of 1,000 people:

10 people have the disease
990 people do not have the disease

So the probability of having the disease is:

$$ P(Disease)=\frac{10}{1000}=1\% $$

Now suppose a medical test is positive.

The test has these properties:

If a person has the disease, the test is positive 90% of the time.
If a person does not have the disease, the test is positive 5% of the time.

```python
# Bayes' Theorem Example

# Probability of having the disease
P_disease = 0.01

# Probability of a positive test if the person has the disease
P_positive_given_disease = 0.90

# Probability of a positive test if the person does not have the disease
P_positive_given_no_disease = 0.05

# Probability of not having the disease
P_no_disease = 1 - P_disease

# Overall probability of a positive test
P_positive = (
    P_positive_given_disease * P_disease
    + P_positive_given_no_disease * P_no_disease
)

# Bayes' Theorem
P_disease_given_positive = (
    P_positive_given_disease * P_disease
) / P_positive

print("Probability of disease given a positive test:",
      P_disease_given_positive)

print("Percentage:",
      P_disease_given_positive * 100, "%")
```

## Probability Rules
Probability rules are mathematical principles used to calculate and combine the probabilities of events. They help us understand the likelihood of individual events, multiple events, complementary events, independent events, dependent events, and conditional events..
### Probability is Between 0 and 1

The probability of any event is always between 0 and 1.

$$ 0 \leq P(A) \leq 1 $$
0 → Event is impossible
1 → Event is certain
Between 0 and 1 → Event has some chance of happening
### Probability of the Sample Space

The probability of all possible outcomes together is always 1.

$$ P(S)=1 $$
Example

For a coin:

$$ S=\{Heads,Tails\} $$

Therefore:

$$ P(Heads)+P(Tails)=1 $$ $$ 0.5+0.5=1 $$

So there is a 100% probability that either Heads or Tails will occur.

$$ P(A)=0 $$

1, 2, 3, 4, 5, 6

Therefore:

$$ P(10)=0 $$

$$ P(A)=1 $$
Example

Rolling a standard dice and getting a number between 1 and 6.

This is guaranteed to happen.

Therefore:

$$ P(1\text{ to }6)=1 $$

## Expected Value
Expected Value is the long-term average value of a random variable, calculated by multiplying each possible outcome by its probability and adding all the results. It helps us understand the average outcome we can expect over many repeated experiments.

E(X)=∑xP(x)
### Example
Suppose you toss a fair coin.

Define:

Heads → ₹100
Tails → ₹0

Each outcome has a probability of 0.5.

Expected value:

$$ E(X)=(100\times0.5)+(0\times0.5) $$ $$ =50+0 $$ $$ \boxed{E(X)=₹50} $$

```python
# Expected Value of rolling a fair dice

values = [1, 2, 3, 4, 5, 6]
probabilities = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]

expected_value = 0

for value, probability in zip(values, probabilities):
    expected_value += value * probability

print("Expected Value:", expected_value)
```
