# Measures of Dispersion
Measures of dispersion describe how much the values in a dataset are spread out or vary from each other and from the central value.

Common Measures of Dispersion
- Range – Difference between the highest and lowest values.
- Variance – Measures the average squared difference from the mean.
- Standard Deviation – Measures the typical distance of values from the mean.
- Quartiles – Divide ordered data into four parts.
- Percentiles – Divide data into 100 parts.
- Interquartile Range (IQR) – Difference between the third and first quartiles.
- Coefficient of Variation (CV) – Measures variation relative to the mean.
### Example

Consider two datasets:

Dataset A: 48, 49, 50, 51, 52
Dataset B: 10, 30, 50, 70, 90

## Range
Range is the simplest measure of dispersion. It tells us the difference between the highest and lowest values in a dataset.
### Formula
$$ \text{Range} = \text{Maximum Value} - \text{Minimum Value} $$
Example

Data:

10, 20, 30, 40, 50

Maximum value = 50
Minimum value = 10
$$ \text{Range} = 50 - 10 = 40 $$

Range = 40

```python
data = [10, 20, 30, 40, 50]

range_value = max(data) - min(data)

print("Range:", range_value)
```


- Small range → values are closer together.
- Large range → values are more spread out.
- Range is easy to calculate but can be strongly affected by outliers.

## Variance
Variance is a measure of dispersion that tells us how far the data values are spread out from the mean.

A small variance means the values are close to the mean, while a large variance means the values are more spread out.



Where:

- x = individual value
- μ = population mean
- N = total number of values
- σ² = population variance
### Business Example
A company compares employee salaries.
If salary variance is high, employee salaries are widely distributed.
If salary variance is low, salaries are relatively close to each other.
### AI/ML Example

In machine learning, variance is used to understand how much a model's predictions can change when the training data changes. It is also important when studying the bias-variance trade-off.
### Example

Data:

2, 4, 6

Step 1: Find the mean

Mean = 2+4+6/3 = 4

Step 2: Find the difference from the mean
| Value | Difference | Squared Difference |
| ----: | ---------: | -----------------: |
|     2 | 2 − 4 = -2 |                  4 |
|     4 |  4 − 4 = 0 |                  0 |
|     6 |  6 − 4 = 2 |                  4 |

Step 3: Calculate variance

Variance = 4+0+4/3

    = 8/3 
        
    = 2.67


```python
import numpy as np

data = [2, 4, 6]

variance = np.var(data)

print("Variance:", variance)
```

## Standard Deviation
Standard deviation measures how far data values are spread out from their mean, and it is the square root of variance.
- Small standard deviation → Values are close to the mean.
- Large standard deviation → Values are widely spread out.
### Formula

### AI/ML Example

Standard deviation is commonly used in data preprocessing and feature scaling. For example, standardization (Z-score scaling) uses the mean and standard deviation to transform features to a comparable scale.
### Example
SD=2.67
​≈1.63

```python
import numpy as np

data = [2, 4, 6]

std = np.std(data)

print("Standard Deviation:", std)
```

## Quartiles
Quartiles are values that divide an ordered dataset into four equal parts.

They help us understand how the data is distributed from the lowest to the highest value.

The Three Quartiles
- Q1 (First Quartile) → 25% of the data is below Q1.
- Q2 (Second Quartile) → 50% of the data is below Q2. It is the median.
- Q3 (Third Quartile) → 75% of the data is below Q3.
Why are Quartiles useful?

Quartiles are useful for:

Understanding data distribution
Identifying outliers
Comparing different datasets
Calculating the Interquartile Range (IQR)
Data analysis and visualization, such as box plots
### Example

Consider the data:

10, 20, 30, 40, 50, 60, 70

Q1 = 20

Q2 = 40

Q3 = 60
| Quartile | Meaning                  | Value |
| -------- | ------------------------ | ----: |
| Q1       | 25th percentile          |    20 |
| Q2       | 50th percentile (Median) |    40 |
| Q3       | 75th percentile          |    60 |
| Quartile | Meaning                  | Value |


```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

## Percentiles
Percentiles are values that divide an ordered dataset into 100 parts. They help us understand the relative position of a value within a dataset.
Important Percentiles
- P25 → 25th percentile
- P50 → 50th percentile (Median)
- P75 → 75th percentile
- P90 → 90th percentile
- P95 → 95th percentile
- P99 → 99th percentile
### Example

Suppose a student is at the 90th percentile in an exam.

This means the student's score is higher than or equal to the scores of approximately 90% of the students in the reference group.

It does not mean the student scored 90%.

10, 20, 30, 40, 50, 60, 70, 80, 90
50th percentile = 50

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80, 90]

p25 = np.percentile(data, 25)
p50 = np.percentile(data, 50)
p90 = np.percentile(data, 90)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("90th Percentile:", p90)
```

Percentiles are commonly used in:

- Exam results
- Salary analysis
- Performance analysis
- Healthcare measurements
- Machine learning and data analysis
- Identifying extreme values

## Interquartile Range (IQR)
Interquartile Range (IQR) is a measure of dispersion that shows the spread of the middle 50% of a dataset.

It is calculated using the first quartile (Q1) and third quartile (Q3).

### Formula
$$ \text{IQR} = Q3 - Q1 $$

Where:

Q1 = First Quartile (25th percentile)
Q3 = Third Quartile (75th percentile)
## Example

Consider the ordered data:

10, 20, 30, 40, 50, 60, 70

- Q1 = 20
- Q2 (Median) = 40
- Q3 = 60


```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Coefficient of Variation
Coefficient of Variation is the standard deviation expressed as a percentage of the mean, used to measure relative variability
### Formula

## AI/ML Example
CV can help compare the relative variability of different numerical features.
For example, two features may have very different units or averages. Comparing their CV values can provide information about their relative dispersion.


```python
import numpy as np

data = [45, 48, 50, 52, 55]

mean = np.mean(data)
std = np.std(data)

cv = (std / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", std)
print("Coefficient of Variation:", cv, "%")
```
