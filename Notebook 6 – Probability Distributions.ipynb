#  Probability Distributions 

## Bernoulli Distribution
### What is Bernoulli Distribution?

Bernoulli Distribution is a probability distribution used when a random experiment has only two possible outcomes.
Bernoulli Distribution 
### Formula

The probability mass function (PMF) is:

$$ \boxed{P(X=x)=p^x(1-p)^{1-x}} $$

Where:

- X = Random variable
- x = 0 or 1
- p = Probability of success
- 1 − p = Probability of failure
#### Example:
Tossing a Coin

Suppose we toss a coin once.

There are only two possible outcomes:

Heads → 1 (Success)
Tails → 0 (Failure)

If the coin is fair:

$$ P(Heads)=0.5 $$ $$ P(Tails)=0.5 $$

This is a Bernoulli distribution because there is only one trial and only two possible outcomes.


```python
import numpy as np

# Probability of success
p = 0.8

# Generate one Bernoulli trial
result = np.random.binomial(n=1, p=p)

print("Result:", result)

if result == 1:
    print("Success")
else:
    print("Failure")
```

## Binomial Distribution
Binomial Distribution is a probability distribution used to find the probability of getting a certain number of successes in a fixed number of independent trials.
#### What is Binomial Distribution?
Binomial Distribution is a discrete probability distribution that calculates the probability of getting a specific number of successes in a fixed number of independent trials, where each trial has two possible outcomes and the probability of success remains constant.

#### Formula

The formula is:

$$ \boxed{ P(X=k)=\binom{n}{k}p^k(1-p)^{n-k} } $$

Where:

- n = Total number of trials
- k = Number of successes we want
- p = Probability of success
- 1 − p = Probability of failure
\(\binom{n}{k}\) = Number of ways to arrange the successes
### Example
Suppose a company contacts 10 customers.
The probability that one customer purchases a product is 0.3.
We can use Binomial Distribution to find the probability that exactly 4 customers purchase the product.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.3

x = np.arange(0, n + 1)

probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

## Poisson Distribution
Poisson Distribution is a discrete probability distribution used to find the probability of a certain number of events happening within a fixed interval of time, distance, area, or space, when those events occur independently at an average rate.
### Formula

The Poisson probability formula is:

$$ P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!} $$

Where:

\(P(X=k)\) = probability of getting exactly k events
\(\lambda\) = average number of events in the interval
\(k\) = number of events we want to find
\(e\) ≈ 2.718
\(k!\) = factorial of k
### Example

Imagine a customer support team receives an average of 4 calls per hour.

We can use Poisson Distribution to find:

- Probability of receiving exactly 2 calls in one hour
- Probability of receiving exactly 4 calls
- Probability of receiving 6 calls
- Probability of receiving no calls

Here, the average number of calls per hour = 4.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 5

x = np.arange(0, 16)

probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

plt.show()
```

## Uniform Distribution

Uniform Distribution is a probability distribution where all possible outcomes have an equal chance of occurring within a given range
There are two types:

Discrete Uniform Distribution – individual values have equal probability.
Continuous Uniform Distribution – every value within a continuous range is equally likely.
### Formula

For a continuous uniform distribution between \(a\) and \(b\):

$$ f(x)=\frac{1}{b-a} $$

Where:

- \(a\) = minimum value
- \(b\) = maximum value
- \(x\) = value within the range
### Example

Imagine a random number generator that can generate any number between:

0 and 10

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 random values between 10 and 20
data = np.random.uniform(10, 20, 1000)

print("First 10 values:")
print(data[:10])

# Plot histogram
plt.hist(data, bins=10)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

## Normal Distribution
Normal Distribution is a continuous probability distribution where most of the data values are concentrated around the mean, and fewer values occur as we move away from the mean.
### Formula

The probability density function of a normal distribution is:

$$ f(x)=\frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$

Where:

- \(x\) = data value
- \(\mu\) = mean
- \(\sigma\) = standard deviation
- \(\pi\) ≈ 3.14159
- \(e\) ≈ 2.718
### Example

Imagine the heights of 1,000 students.

Most students may have heights around:

160 cm – 175 cm

Only a few students may be:

140 cm
200 cm


```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 10

data = np.random.normal(mean, std, 10000)

plt.hist(data, bins=30, density=True)

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")

plt.show()
```

## Exponential Distribution
Exponential Distribution is a continuous probability distribution used to model the waiting time until an event happens.
### Formula

The probability density function of Exponential Distribution is:

$$ f(x)=\lambda e^{-\lambda x} $$

Where:

- \(x\) = waiting time
- \(\lambda\) = average event rate
- \(e\) ≈ 2.718
#### Example

Imagine a customer support center receives an average of 6 calls per hour.

We can use Exponential Distribution to answer questions like:

How long will we wait for the next call?
What is the probability that the next call arrives within 5 minutes?
What is the probability that we wait more than 10 minutes?

```python
import numpy as np
import matplotlib.pyplot as plt

# Average event rate
lambda_value = 6

# Generate 1000 waiting times
waiting_times = np.random.exponential(
    scale=1 / lambda_value,
    size=1000
)

# Display first 10 values
print("First 10 waiting times:")
print(waiting_times[:10])

# Plot the distribution
plt.hist(waiting_times, bins=30)

plt.xlabel("Waiting Time (hours)")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")

plt.show()
```
