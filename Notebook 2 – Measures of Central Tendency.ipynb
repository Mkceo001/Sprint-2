# Notebook 2 – Measures of Central Tendency

## What are Measures of Central Tendency?
Measures of central tendency are statistical values that describe the center or typical value of a dataset.

The main measures covered in this notebook are:
1. Mean
2. Median
3. Mode
4. Weighted Mean
5. Geometric Mean
6. Harmonic Mean
## Mean
Mean is the average value of a set of numbers. It is one of the most common measures of central tendency.
### Formula


- (x\) = each value
- (n\) = number of values
- (sum x\) = sum of all values

Mean is the value obtained by dividing the sum of all observations by the total number of observations.
### Example
Mean = 60+70+80+90+100
       

```python
import numpy as np

data = [60, 70, 80, 90, 100]

mean_value = np.mean(data)

print("Mean:", mean_value)
```

## Median
Median is the middle value of a dataset when all values are arranged in ascending or descending order
### Formula

### example:
Even number of values

Find the middle of that data

Data: 10, 20, 30, 40

There are two middle values: 20 and 30

Median= 20+30 / 2


```python
import numpy as np

data = [10, 20, 30, 40, 50]

median_value = np.median(data)

print("Median:", median_value)
```

## Mode
Mode is the value that appears most frequently in a dataset.
A dataset can have:
- One mode → Unimodal
- Two modes → Bimodal
- Multiple modes → Multimodal
- No mode → When no value repeats
### Example

Data: 10, 20, 20, 30, 40, 20, 50

Here, 20 appears three times, which is more than any other value.

Mode = 20

Another Example

Data: 5, 10, 15, 15, 20, 25, 15

15 occurs most often.

Mode = 15

```python
from collections import Counter

data = [ 5, 10, 15, 15, 20, 25, 15]

counts = Counter(data)

mode_value = counts.most_common(1)[0][0]

print("Mode:", mode_value)
```

## Weighted Mean
Weighted Mean is an average where some values have more importance (weight) than others.

Instead of treating every value equally, we give different weights based on their importance.
### Formula


Where:

x = value
w = weight
### Example
| Assessment | Score | Weight |
|---|---:|---:|
| Assignment | 80 | 20% |
| Mid Exam | 70 | 30% |
| Final Exam | 90 | 50% |

(80 x 0.20)+(70+0.30)+(90+0.50)

=16+21+45

=82

```python
scores = [80, 70, 90]
weights = [0.20, 0.30, 0.50]

weighted_mean = sum(
    score * weight
    for score, weight in zip(scores, weights)
) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

## Geometric Mean  
The geometric mean is an average that is calculated by multiplying values together and then taking the nth root.

Geometric Mean is commonly used for:

- Growth rates
- Investment returns
- Population growth
- Financial analysis
- Ratios and percentages
  
### Formula


### Example

Data: 2, 4, 8

Step 1: Multiply the values

$$ 2 \times 4 \times 8 = 64 $$

Step 2: Take the cube root because there are 3 values.

GM = 3 root of 64 = 4

### Example

```python
import numpy as np

data = [1.10, 1.20, 1.30]

geometric_mean = np.prod(data) ** (1 / len(data))

print("Geometric Mean:", geometric_mean)
```

## Harmonic Mean
Harmonic Mean (HM) is a type of average that is useful when dealing with rates, ratios, speeds, and other quantities expressed per unit.

Harmonic Mean is commonly used for:

- Average speed
- Rates
- Ratios
- Financial calculations
- Performance measurements
### Formula



### Example:

Speeds= 40 km/h, 60 km/h

Calculation:

HM = 2 / (1/40 + 1/60)

HM = 48 km/h

Precision = 60
Recall = 40


```python
data = [60, 40]

harmonic_mean = len(data) / sum(1 / x for x in data)

print("Harmonic Mean:", harmonic_mean)
```
