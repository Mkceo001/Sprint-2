## Explain the difference between Mean and Median.?
Mean

The Mean is the average of all values.

Formula
$$ Mean = \frac{\text{Sum of all values}}{\text{Number of values}} $$
Example

Dataset:

10, 20, 30, 40, 50
Mean=
5
10+20+30+40+50
	​

$$ Mean = 30 $$
2. Median

The Median is the middle value after arranging the data in ascending or descending order.

For an odd number of values:

10, 20, 30, 40, 50

The middle value is:

30

So:

$$ Median = 30 $$

For an even number of values, we take the average of the two middle values.

Example:

10, 20, 30, 40
$$ Median = \frac{20+30}{2}=25 $$

```python
data = [10, 20, 30, 40, 500]

mean = sum(data) / len(data)

print("Mean:", mean)
```

```python
data = [10, 20, 30, 40, 500]

data.sort()

n = len(data)

if n % 2 == 1:
    median = data[n // 2]
else:
    median = (data[n // 2 - 1] + data[n // 2]) / 2

print("Median:", median)
```

## What is Variance?
Answer:
Variance measures how far the data values are spread from the mean.
A low variance means values are close to the mean.
A high variance means values are widely spread.
#### Formula:

### Example

Suppose two students have these marks:

Student A: 48, 49, 50, 51, 52
Student B: 20, 35, 50, 65, 80

```python
import numpy as np

student_a = [48, 49, 50, 51, 52]
student_b = [20, 35, 50, 65, 80]

print("Student A Variance:", np.var(student_a))
print("Student B Variance:", np.var(student_b))
```

## Explain Bayes' Theorem with an example.
Answer:
Bayes' Theorem calculates the probability of an event based on prior knowledge.
#### Formula:
P(A∣B)=P(B)P(B∣A)×P(A)​


```python
# Given probabilities
p_spam = 0.20
p_not_spam = 0.80

p_offer_given_spam = 0.60
p_offer_given_not_spam = 0.10

# Calculate P(Offer)
p_offer = (
    p_offer_given_spam * p_spam
    + p_offer_given_not_spam * p_not_spam
)

# Bayes' Theorem
p_spam_given_offer = (
    p_offer_given_spam * p_spam
) / p_offer

print("Probability of Offer:", p_offer)
print("Probability of Spam given Offer:", p_spam_given_offer)
print("Percentage:", p_spam_given_offer * 100, "%")
```

## Explain Correlation and Covariance.?
Covariance shows whether two variables move together.
- Positive covariance → variables tend to increase together.
- Negative covariance → one tends to increase when the other decreases.
Correlation measures both the direction and strength of the relationship.

Correlation ranges from -1 to +1.
- +1 → Perfect positive relationship
- 0 → No linear relationship
- -1 → Perfect negative relationship

```python
import numpy as np

exercise = [1, 2, 3, 4, 5]
weight = [80, 77, 74, 70, 67]

correlation = np.corrcoef(exercise, weight)[0, 1]

print("Correlation:", correlation)
```

## What is IQR?

IQR (Interquartile Range) is a statistical measure that tells us the spread of the middle 50% of the data.

In simple words:

IQR is the difference between the third quartile (Q3) and the first quartile (Q1).

Formula
$$ \boxed{IQR = Q3 - Q1} $$

Where:

Q1 = 25th percentile
Q2 = 50th percentile (Median)
Q3 = 75th percentile

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 1000]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = [
    x for x in data
    if x < lower_bound or x > upper_bound
]

print("Outliers:", outliers)
```

## What is a p-value?
Answer:
A p-value helps determine whether there is enough statistical evidence to reject the null hypothesis.
Common rule:
- p-value < 0.05 → Reject the null hypothesis.
- p-value ≥ 0.05 → Fail to reject the null hypothesis.

```python
from scipy.stats import ttest_1samp

# Employee performance scores
scores = [52, 55, 54, 58, 60, 57, 59, 61, 56, 58]

# Test whether the average score is different from 50
t_stat, p_value = ttest_1samp(scores, 50)

print("t-statistic:", t_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("The result is statistically significant.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("The result is not statistically significant.")
```

## Detect Outliers using the IQR Method.?

```python
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 100]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [
    x for x in data
    if x < lower_bound or x > upper_bound
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```
